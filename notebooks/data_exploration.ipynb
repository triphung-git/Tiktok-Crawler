```python
import sys
from pathlib import Path

# Add project root to sys.path
PROJECT_ROOT = Path("..").resolve()
sys.path.insert(0, str(PROJECT_ROOT))

from src.config import get_config, resolve_path
from src.storage.json_storage import load_json, GlobalIndexManager

config = get_config()
print("Project Root:", PROJECT_ROOT)
print("Pipeline Name:", config.get("pipeline", {}).get("name"))
```

```python
# Load and inspect global processed index
index_mgr = GlobalIndexManager()
index_data = index_mgr.load()
print(f"Total unique video URLs tracked: {index_data.get('total_records', 0)}")
print(f"Last updated at: {index_data.get('updated_at')}")
```

```python
# Inspect dialect breakdown from tasks
from collections import Counter

tasks_files = sorted(resolve_path("data/processed/audio").glob("**/sources_*.json"))
if tasks_files:
    sample_file = tasks_files[-1]
    tasks = load_json(sample_file, default=[])
    print(f"Loaded {len(tasks)} tasks from {sample_file.name}")
    dialects = Counter(t.get("language_region", "unknown") for t in tasks)
    print("Dialect distribution:", dict(dialects))
else:
    print("Chưa có file sources_*.json nào trong data/processed/audio/")
```
